```python
start = time.time()

N = 1000
x = np.linspace(0, 100, N)
a = np.zeros(N)

for i in range(N):
    for j in range(N):
        if i != j:
            b = (x[i] - x[j]) ** -2
            a[i] = a[i] + b
            
end = time.time()

print (end-start)
```

```python
start = time.time()

N = 1000
x = np.linspace(0, 100, N)
a = np.zeros(N)

for i in range(N):
    for j in range(i):
        b = (x[i] - x[j]) ** -2
        a[i] = a[i] + b
    for j in range(i+1, N):
        b = (x[i] - x[j]) ** -2
        a[i] = a[i] + b
end = time.time()

print (end-start)
```
